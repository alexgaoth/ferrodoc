## Loading

The extract is one row per order. `read_csv` infers the types, which
is wrong for the postcode column:

> Postcodes are strings. A postcode that starts with a zero loses it
> the moment anything treats it as a number.

- `order_id` — integer
- `postcode` — string
- `total` — float
